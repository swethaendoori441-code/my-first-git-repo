```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("test").getOrCreate()
spark
```

```python
data = [
    (101, "Electronics", "USA", 500),
    (102, "Electronics", "USA", 300),
    (103, "Clothing", "UK", 100),
    (104, "Clothing", "USA", 150),
    (105, "Home", "UK", 200),
    (106, "Home", "India", 400),
]
columns = ["product_id", "category", "country","price"]
df = spark.createDataFrame(data,columns)
df.show()
```

```python
df.write.mode("overwrite").partitionBy("country").parquet("/content/partition")
```

```python
df_india = spark.read.parquet("/content/partition/country=India/part-00001-15a6e0dd-e5c9-4217-836c-2bb5bb938d77.c000.snappy.parquet")
df_india.show()
```
